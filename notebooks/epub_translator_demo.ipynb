## ⚙️ Configuration

Configure your translation settings.

````python
# Translation settings
SOURCE_LANG = 'en'      # Source language (e.g., 'en', 'fr', 'de')
TARGET_LANG = 'fr'      # Target language (e.g., 'en', 'fr', 'de')
MODEL_TYPE = 'marian'   # Options: 'marian', 'm2m100', 'nllb'
OUTPUT_FILE = 'translated.epub'

# Advanced settings
BATCH_SIZE = 8          # Increase if you have more GPU memory
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

print("Configuration:")
print(f"  Source language: {SOURCE_LANG}")
print(f"  Target language: {TARGET_LANG}")
print(f"  Model type: {MODEL_TYPE}")
print(f"  Device: {DEVICE}")
print(f"  Batch size: {BATCH_SIZE}")
````

## 🚀 Start Translation

Now let's translate your ePub!

````python
from epub_translator import translate_epub
import time

print("Starting translation...\n")
start_time = time.time()

try:
    translate_epub(
        input_path=input_file,
        output_path=OUTPUT_FILE,
        source_lang=SOURCE_LANG,
        target_lang=TARGET_LANG,
        model_type=MODEL_TYPE,
        device=DEVICE,
        batch_size=BATCH_SIZE,
    )
    
    elapsed_time = time.time() - start_time
    print(f"\n✓ Translation completed successfully!")
    print(f"  Time elapsed: {elapsed_time / 60:.2f} minutes")
    print(f"  Output file: {OUTPUT_FILE}")
    
except Exception as e:
    print(f"\n✗ Error during translation: {e}")
    import traceback
    traceback.print_exc()
````

## 📖 Examples & Tips

### Different Model Types

**MarianMT** (Fast, ~50 language pairs)
```python
MODEL_TYPE = 'marian'
```

**M2M100** (Good balance, 100 languages)
```python
MODEL_TYPE = 'm2m100'
```

**NLLB** (Best quality, 200+ languages)
```python
MODEL_TYPE = 'nllb'
```

### Common Language Codes

- `ru` - Russian
- `zh` - Chinese
- `ja` - Japanese
- `ar` - Arabic

### Performance Tips

1. **Enable GPU** for 5-10x speedup (Runtime > Change runtime type > GPU)
2. **Increase batch size** if you have GPU memory (try 16 or 32)
3. **Use MarianMT** for faster translations
4. **Use NLLB** for best quality

### Troubleshooting

**Out of Memory?**
- Reduce batch size to 4 or 2
- Use smaller model (MarianMT instead of NLLB)

**Slow Translation?**
- Enable GPU in Colab settings
- Use MarianMT model
- Increase batch size

## 🎓 Advanced Usage

### Custom Model

````python
# Use a specific model for best quality
translate_epub(
    input_path=input_file,
    output_path='output_nllb.epub',
    source_lang='en',
    target_lang='fr',
    model_type='nllb',
    model_name='facebook/nllb-200-distilled-600M',
    device='cuda',
    batch_size=16,
)
````

### Low-Level API

````python
from epub_translator import EPubParser, Translator, TextProcessor

# Parse ePub
parser = EPubParser(input_file)
parser.load()
parser.extract_translatable_content()

# Get texts
texts = parser.get_all_texts()
print(f"Found {len(texts)} text elements")

# Process texts
processor = TextProcessor(max_length=512)
chunks, mapping = processor.chunk_texts(texts)
print(f"Split into {len(chunks)} chunks")

# Translate
translator = Translator(
    source_lang='en',
    target_lang='fr',
    model_type='marian'
)
translated_chunks = translator.translate(chunks)

# Reconstruct
translations = processor.reconstruct_texts(translated_chunks, mapping, len(texts))

# Update and save
parser.update_with_translations(translations)
parser.save('custom_output.epub')

print("✓ Done!")
````
