## 📤 ÉTAPE 3 : Upload de votre fichier ePub

## ⚙️ ÉTAPE 4 : Configuration de la traduction

Choisissez le modèle de traduction. **Trois options disponibles :**

| Modèle | Qualité | Vitesse | Utilisation |
|--------|---------|---------|-------------|
| **marian** | ⭐⭐⭐ | 🚀🚀🚀 Rapide | Romans, textes simples |
| **m2m100** | ⭐⭐⭐⭐ | 🚀🚀 Moyen | Usage général |
| **nllb** | ⭐⭐⭐⭐⭐ | 🚀 Plus lent | Textes techniques, meilleure qualité |

**Recommandation : Commencez avec `marian` pour tester, puis `nllb` pour qualité maximale.**

````python
# ===== CONFIGURATION =====

# Langues source et cible
SOURCE_LANG = 'en'  # Anglais
TARGET_LANG = 'fr'  # Français

# Choisissez votre modèle (changez cette ligne selon vos besoins)
MODEL_TYPE = 'marian'  # Options : 'marian', 'm2m100', 'nllb'

# Nom du fichier de sortie
OUTPUT_FILE = 'livre_traduit_fr.epub'

# ===== PARAMÈTRES AVANCÉS (laisser par défaut) =====

# Taille du batch (augmentez si vous avez assez de GPU)
BATCH_SIZE = 16 if torch.cuda.is_available() else 4

# Device
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# ===== AFFICHAGE =====

print("\n" + "="*50)
print("⚙️  CONFIGURATION DE LA TRADUCTION")
print("="*50)
print(f"\n📖 Langue source    : {SOURCE_LANG} (Anglais)")
print(f"🎯 Langue cible     : {TARGET_LANG} (Français)")
print(f"🤖 Modèle utilisé   : {MODEL_TYPE.upper()}")
print(f"💾 Fichier sortie   : {OUTPUT_FILE}")
print(f"⚡ Device           : {DEVICE.upper()}")
print(f"📦 Taille du batch  : {BATCH_SIZE}")

if MODEL_TYPE == 'marian':
    print("\n💡 Vous utilisez MarianMT - Traduction rapide !")
elif MODEL_TYPE == 'm2m100':
    print("\n💡 Vous utilisez M2M100 - Bon équilibre qualité/vitesse")
elif MODEL_TYPE == 'nllb':
    print("\n💡 Vous utilisez NLLB - Meilleure qualité (plus lent)")

print("\n" + "="*50)
print("✅ Configuration prête !")
print("="*50)
````

## 🚀 ÉTAPE 5 : Lancer la traduction !

C'est parti ! La traduction va commencer. **Cela peut prendre plusieurs minutes selon la taille du livre.**

### Temps estimés (livre de 300 pages) :
- 🚀 Avec GPU + MarianMT : **5-10 minutes**
- 🚀 Avec GPU + NLLB : **15-25 minutes**
- 🐌 Sans GPU : **30-60 minutes** (activez le GPU !)

**Une barre de progression s'affichera pendant la traduction.**

````python
from epub_translator import translate_epub
import time
from datetime import datetime

if input_file is None:
    print("❌ Erreur : Aucun fichier uploadé. Retournez à l'ÉTAPE 3.")
else:
    print("\n" + "="*60)
    print("🚀 DÉMARRAGE DE LA TRADUCTION")
    print("="*60)
    print(f"\n⏰ Heure de début : {datetime.now().strftime('%H:%M:%S')}")
    print(f"📖 Fichier : {input_file}")
    print(f"🎯 Traduction : Anglais → Français\n")
    print("-" * 60)
    
    start_time = time.time()
    
    try:
        # Lancer la traduction
        translate_epub(
            input_path=input_file,
            output_path=OUTPUT_FILE,
            source_lang=SOURCE_LANG,
            target_lang=TARGET_LANG,
            model_type=MODEL_TYPE,
            device=DEVICE,
            batch_size=BATCH_SIZE,
        )
        
        # Calcul du temps écoulé
        elapsed_time = time.time() - start_time
        elapsed_minutes = elapsed_time / 60
        
        # Succès !
        print("\n" + "="*60)
        print("🎉 TRADUCTION TERMINÉE AVEC SUCCÈS !")
        print("="*60)
        print(f"\n⏱️  Temps écoulé : {elapsed_minutes:.1f} minutes ({elapsed_time:.0f} secondes)")
        print(f"⏰ Heure de fin : {datetime.now().strftime('%H:%M:%S')}")
        print(f"📁 Fichier de sortie : {OUTPUT_FILE}")
        print(f"\n✅ Votre livre est prêt en français !")
        print("\n👉 Passez à l'ÉTAPE 6 pour le télécharger.")
        print("="*60)
        
    except Exception as e:
        print("\n" + "="*60)
        print("❌ ERREUR PENDANT LA TRADUCTION")
        print("="*60)
        print(f"\n{str(e)}\n")
        
        # Suggestions de résolution
        print("💡 Solutions possibles :")
        print("\n1. Si erreur 'Out of Memory' :")
        print("   - Réduisez BATCH_SIZE à 4 ou 2")
        print("   - Utilisez MODEL_TYPE = 'marian' (plus léger)")
        print("\n2. Si erreur de téléchargement de modèle :")
        print("   - Vérifiez votre connexion internet")
        print("   - Réessayez dans quelques minutes")
        print("\n3. Pour plus de détails :")
        import traceback
        traceback.print_exc()
````

## 📥 ÉTAPE 6 : Télécharger votre livre traduit

Téléchargez votre livre ePub traduit en français !

````python
from google.colab import files
import os

print("📥 Téléchargement du livre traduit...\n")

if os.path.exists(OUTPUT_FILE):
    file_size_mb = os.path.getsize(OUTPUT_FILE) / 1024 / 1024
    
    print(f"✅ Fichier trouvé : {OUTPUT_FILE}")
    print(f"📊 Taille : {file_size_mb:.2f} MB\n")
    print("⬇️  Le téléchargement va démarrer...\n")
    
    files.download(OUTPUT_FILE)
    
    print("\n" + "="*60)
    print("🎉 TÉLÉCHARGEMENT TERMINÉ !")
    print("="*60)
    print("\n✅ Votre livre traduit est maintenant sur votre ordinateur !")
    print("\n📚 Profitez de votre lecture en français !")
    print("\n💡 Vous pouvez lire les fichiers .epub avec :")
    print("   - Calibre (PC/Mac/Linux)")
    print("   - Apple Books (Mac/iPhone/iPad)")
    print("   - Google Play Books (Android)")
    print("   - Adobe Digital Editions")
    print("\n" + "="*60)
else:
    print("❌ Erreur : Fichier de sortie non trouvé.")
    print("\n💡 Assurez-vous que l'ÉTAPE 5 (traduction) a réussi.")
````

# 🎓 BONUS : Fonctionnalités Avancées

## Traduire plusieurs livres d'un coup

````python
# Upload multiple files
from google.colab import files

print("📤 Uploadez plusieurs fichiers ePub...\n")
uploaded_files = files.upload()

if uploaded_files:
    print(f"\n✅ {len(uploaded_files)} fichier(s) uploadé(s)\n")
    
    for i, filename in enumerate(uploaded_files.keys(), 1):
        print(f"\n{'='*60}")
        print(f"📖 Traduction {i}/{len(uploaded_files)} : {filename}")
        print('='*60)
        
        output_name = filename.replace('.epub', '_fr.epub')
        
        try:
            translate_epub(
                input_path=filename,
                output_path=output_name,
                source_lang='en',
                target_lang='fr',
                model_type='marian',  # Rapide pour batch
                device='cuda' if torch.cuda.is_available() else 'cpu',
                batch_size=16
            )
            print(f"✅ {filename} → {output_name}")
            
            # Télécharger immédiatement
            files.download(output_name)
            
        except Exception as e:
            print(f"❌ Erreur pour {filename}: {e}")
    
    print(f"\n{'='*60}")
    print("🎉 Traduction batch terminée !")
    print('='*60)
````

```python

    input_path=input_file,

    source_lang='en',
```

```python
    model_type='marian',
    device='cuda',

```

```python
# Réduisez le batch size

    source_lang='en',
    target_lang='fr',
    model_type='marian',
    device='cuda',
    batch_size=2,  # Plus petit = moins de mémoire
    max_chunk_length=256  # Chunks plus petits
)
```
